```python
import glob
import numpy as np
import pandas as pd

files = sorted(glob.glob("../data_raw/SDR-*.csv"))
cols_to_keep = [
    'OperatorControlNumber', 'DifficultyDate', 'SubmissionDate',
    'OperatorDesignator', 'SDRType', 'JASCCode', 'NatureOfConditionA',
    'PrecautionaryProcedureA', 'StageOfOperationCode', 'HowDiscoveredCode',
    'RegistryNNumber', 'AircraftMake', 'AircraftModel', 'AircraftSerialNumber',
    'AircraftTotalTime', 'AircraftTotalCycles', 'PartName', 'PartCondition',
    'PartLocation', 'Discrepancy'
]

df = pd.concat([pd.read_csv(f, usecols=cols_to_keep, low_memory=False) for f in files], ignore_index=True)

# تحديد صيغة التاريخ مباشرة لتسريع التنفيذ 20 مرة
df['DifficultyDate'] = pd.to_datetime(df['DifficultyDate'], format='%m/%d/%Y', errors='coerce')
df['SubmissionDate'] = pd.to_datetime(df['SubmissionDate'].str[:10], format='%Y-%m-%d', errors='coerce')

df = df.dropna(subset=['DifficultyDate'])
df['Reporting_Lag_Days'] = (df['SubmissionDate'] - df['DifficultyDate']).dt.days.clip(lower=0)

print("تم الدمج وتنظيف التواريخ بسرعة! عدد الصفوف السليمة:", f"{len(df):,}")
```
